# Exploración de experimentos

Índice: una fila por run, una columna por atributo (`autoplots/index.py`). Las mismas consultas que `aplot --where`.

```python
import os, sys
os.environ.setdefault('JAX_PLATFORMS', 'cpu')
sys.path.insert(0, os.path.abspath('..'))
%matplotlib inline
from autoplots.index import load_index, select, ensure, column
from autoplots.aplot import main as aplot

df = load_index(['../Simulations_claude_cesga', '../Simulations_claude_garmonegro'])
df.columns.tolist()
```

```python
sel = select(df, "L==8 and status=='ok' and not vscore>0.05")
sel[['job', 'irrep', 'dtype', 'E_site_spin', 'E50', 'vscore', 's_it']].sort_values('E50')
```

```python
# pandas libre: agrupar, pivotar...
sel.pivot_table(index='irrep', columns='dtype', values='E_site_spin', aggfunc='min')
```

```python
# la misma figura que en la terminal (se muestra aquí con --show)
aplot("y=E_site_spin x=irrep hue=dtype --where 'L==8' --show -o Figures")
```

```python
aplot("trace y=E_site_spin smooth=50 sched=1 --where \"job~='c8_CvT3'\" --show -o Figures")
```
